## Grouby Operations

Pandas has some very extensive methods and functions for performing many high-level operations. One really useful one is the `groupby()` method. Here we have subset our data for a decade, but what if we wanted to know the average maximum temperature over that decade?

If we only stored our data in a NumPy array, we would have to do some creative subsetting to get the right data. We would likely be more than a little flummoxed by having a few leap years in there, which would throw off any easy methods for getting the right data.

The Pandas `groupby` method allows us to group by one or more columns. So to group by the day of the year, we would need to group by both the month and the day. Here, we can exploit our index values being dates and use them to pull both the month and the day to perform our `groupby()` operations.

To get the month: `df.index.month`

To get the day: `df.index.day`

This is similar to working with a `datetime` object and getting out the individual pieces of information. Here, we just access our DataFrame's index values using the "dot" method, and since those are datetime-like objects, we can simply access the appropriate attributes.

To perform the group by for a single value:

```python
df.groupby(value1)
```

or for two variables:

```python
df.groupby([value1, value2])
```

Where `value1` and `value2` are each arrays that represent the row information you wish to group by.

````python
gb = ltm_data.groupby([])
````
