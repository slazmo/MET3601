### For more on the NAO and what it means, visit:
https://www.climate.gov/news-features/understanding-climate/climate-variability-north-atlantic-os




The North Atlantic Oscillation (NAO) index is based on the surface **sea-level pressure difference** between the Subtropical (Azores) High and the Subpolar Low. 

-  **Positive Phase (NAO+):** The pressure difference between the high and low is large. The Azores High strengthens and the Icelandic Low deepens, creating a strong pressure gradient that accelerates westerly winds. This brings mild, wet winters to Northern Europe and warmer conditions to the U.S. East Coast.

- **Negative Phase (NAO–):** The pressure difference is weak. Both the Azores High and Icelandic Low are weaker than normal, slackening the westerly winds. This allows cold Arctic air to plunge south into the eastern United States and Europe, leading to increased snow and storminess.

In the average state of the atmosphere, the North Atlantic surface pressure is relatively high in the subtropics at latitudes 20°N to 40°N ('the Azores High'), and lower further North at latitudes 50°N to 70°N (the 'Icelandic Low'). The North-South pressure difference determines the strength of the westerly winds across the Atlantic and is known as the North Atlantic Oscillation (NAO).

When the pressure difference is large, the NAO is positive and the westerly winds are strong and storms tend to be stronger, more frequent and travel across northwestern Europe. When the pressure difference is small, they travel across southern Europe. The NAO is also associated with changes in temperature and rainfall in Europe and North America.

Statistical Functions: http://docs.scipy.org/doc/numpy/reference/routines.statistics.html
* `corrcoef`
* `median`
* `mean`
* `average`
* `nanmean`
* `max`

A number of these functions can also be called as methods on a Numpy data object.

```python
jan_avg = np.mean(data[::12, 1])
jan_avg_method = data[::12, 1].mean()
print(jan_avg)
print(jan_avg_method)
```
Output:
```
0.12416666666666669
0.12416666666666669
```
Note: The `average` function computes a weighted average (with default weights of 1 for all observations) and the `mean` function just computes the straight arithmatic mean of the array along an axis.

Each of these Numpy functions can be computed over the entire array and return a single value, or over a particular axis to return the average for each row or column depending on which axis was chosen. To do this you would specify the keyword argument of `axis` and set it to the desired axis number (e.g., 0, 1, 2, etc.)
