````python
from datetime import datetime, timezone
from zoneinfo import ZoneInfo

utc_now = datetime.now(timezone.utc)
et_now  = utc_now.astimezone(ZoneInfo("America/New_York"))

# %I = 12-hour hour, %M = minute, %p = AM/PM
formatted_date = et_now.strftime("%A, %B %d, %Y")
formatted_time = et_now.strftime("%I:%M %p")

print(f"Current Date: {formatted_date}")
print(f"The time is:  {formatted_time}")
````
